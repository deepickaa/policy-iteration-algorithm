```python
import gymnasium as gym
import numpy as np

env = gym.make("FrozenLake-v1", is_slippery=True)

def generate_trajectory(policy, max_steps=100):

    trajectory = []

    state, _ = env.reset()

    for _ in range(max_steps):

        action = policy[state]

        next_state, reward, terminated, truncated, _ = env.step(action)

        trajectory.append((state, action, reward))

        state = next_state

        if terminated or truncated:
            break

    return trajectory
```

```python
def mc_prediction(policy, env, num_episodes, gamma=0.99):

    n_states = env.observation_space.n

    returns_sum = {s: 0.0 for s in range(n_states)}
    returns_count = {s: 0 for s in range(n_states)}

    V = np.zeros(n_states)

    for episode in range(num_episodes):

        trajectory = generate_trajectory(policy)

        visited_states = set()

        G = 0

        for t in reversed(range(len(trajectory))):

            state, action, reward = trajectory[t]

            G = gamma * G + reward

            if state not in visited_states:

                visited_states.add(state)

                returns_sum[state] += G
                returns_count[state] += 1

                V[state] = (
                    returns_sum[state] / returns_count[state]
                )

    return Vimport numpy as np

def decay_schedule(initial_value, min_value, decay_ratio, max_steps):

    decay_steps = int(max_steps * decay_ratio)

    values = []

    for step in range(max_steps):

        if step < decay_steps:

            value = initial_value - (
                (initial_value - min_value)
                * (step / decay_steps)
            )

        else:
            value = min_value

        values.append(value)

    return np.array(values)
```

```python
def policy_evaluation(policy, env, gamma=0.99, theta=1e-6):

    n_states = env.observation_space.n

    V = np.zeros(n_states)

    while True:

        delta = 0

        for s in range(n_states):

            v = 0

            for prob, next_state, reward, done in env.P[s][policy[s]]:

                v += prob * (reward + gamma * V[next_state])

            delta = max(delta, abs(v - V[s]))

            V[s] = v

        if delta < theta:
            break

    return V
```

```python
import gymnasium as gym
import numpy as np

env = gym.make("FrozenLake-v1", is_slippery=True)

# Random Policy
policy = np.random.choice(4, size=env.observation_space.n)

# Monte Carlo Prediction
mc_values = mc_prediction(
    policy,
    env,
    num_episodes=5000
)

# Dynamic Programming
dp_env = gym.make(
    "FrozenLake-v1",
    is_slippery=True
).unwrapped

dp_values = policy_evaluation(
    policy,
    dp_env
)

print("\nMonte Carlo State Values:\n")
print(mc_values.reshape(4,4))

print("\nDynamic Programming State Values:\n")
print(dp_values.reshape(4,4))
```
